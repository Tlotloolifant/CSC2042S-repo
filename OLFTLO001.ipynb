Import libaries 

```python

import numpy as np
import matplotlib as mp
from PIL import Image 
import os 



```

Loading images 

```python
data_path = "dataset"

#file path grayscale train and test sets 
grayscle_train_path = os.path.join(data_path, "grayscale",'train')
grayscle_test_path = os.path.join(data_path, "grayscale",'test')


#file path for rgb train and test sets
rgb_train_path= os.path.join(data_path, "rgb",'train')
rgb_test_path= os.path.join(data_path, "rgb",'test')

```

```python
#define classes and labels 

class_names = ["bart_simpson", 
               "charles_montgomery_burns", 
               "homer_simpson", 
               "krusty_the_clown",
               "lisa_simpson", 
               "marge_simpson", 
               "milhouse_van_houten", 
               "moe_szyslak", 
               "ned_flanders",
               "principal_skinner"]

#dictionary to map character/class to a number/label
class_to_label ={}
for index, name in enumerate(class_names):
    class_to_label[name] = index
```

```python
#loading function

def load_images(path):
    x= []
    y =[]
    for class_name in class_names:
#assume that the path inputed is the folder JUST before you enter the class folder/ folder with images 
        class_path = os.path.join(path, class_name)

        #currently holds the path to a specific class (character folder)

        label = class_to_label[class_name]
        # label variable to hold the corresponding number to the relevant class/character

        for image in os.listdir(class_path):
            #for picture in each character folder

            image_path = os.path.join(class_path, image)
            #path to specific indivdual images 

            image = Image.open(image_path)
            #open image

            image_array = np.array(image)
            #convert image into array and flatten

            x.append(image_array.flatten())
            y.append(label)

            
    return np.array(x), np.array(y)
```

```python
#load data

x_gray_train, y_gray_train = load_images(grayscle_train_path)
x_gray_test, y_gray_test = load_images(grayscle_test_path)

x_rgb_train, y_rgb_train = load_images(rgb_train_path)
x_rgb_test, y_rgb_test = load_images(rgb_test_path)


```
